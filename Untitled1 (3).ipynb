```python
import pandas as pd

# File .xls nu irunthalum ulla csv maadhiri irukkum, athan
try:
    df = pd.read_excel('ecommerce_sales_cleaned_data (1) (1).xls')
except:
    df = pd.read_csv('ecommerce_sales_cleaned_data (1) (1).xls')

df.head()
```

```python
print("Total Rows & Columns:", df.shape)
df.info()
```

```python
import matplotlib.pyplot as plt

df['City'].value_counts().plot(kind='bar', color='skyblue')
plt.title('City Wise Orders')
plt.show()
```

```python
df['Gender'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Male vs Female Customers')
plt.show()
```

```python
print(df.columns.tolist())
```

```python
# Un dataset la Product_ID than irukku, atha use pannalam da
df['Product_ID'].value_counts().head(5).plot(kind='bar', color='orange')
plt.title('Top 5 Products')
plt.show()
```

```python
# Sales amount column name check pannu da
df['Shipping_Cost'].describe()
```

```python
df['Region'].value_counts().plot(kind='bar', color='green')
plt.title('Region Wise Sales - TN')
plt.show()
```

```python
df['Age'].hist(bins=20, color='purple')
plt.title('Customer Age Distribution')
plt.xlabel('Age')
plt.show()
```

```python
print("Total Orders:", len(df))
print("Avg Shipping Cost:", df['Shipping_Cost'].mean())
print("Top City:", df['City'].mode()[0])
print("Most Common Region:", df['Region'].mode()[0])
```
